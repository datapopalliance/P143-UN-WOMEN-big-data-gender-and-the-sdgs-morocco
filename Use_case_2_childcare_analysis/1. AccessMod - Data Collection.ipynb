```python
# define functions
# downloading
def download_file(url, path):
    try:
        response = requests.get(url, allow_redirects=True)
        response.raise_for_status()  # Raise an error for bad status codes
        with open(path, 'wb') as file:
            file.write(response.content)
        print(f'{os.path.basename(path)} saved.')
    except requests.RequestException as e:
        print(f'Error downloading {url}: {e}')

# unzipping
def unzip_file(zip_path, extract_to):
    try:
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_to)
        print(f'{os.path.basename(zip_path)} unzipped successfully.')
    except zipfile.BadZipFile as e:
        print(f'Error unzipping {zip_path}: {e}')
```

### 1.1. Administrative Boundaries

Data source: [HDX: UN OCHA : Bangladesh - Subnational Administrative Boundaries ](https://data.humdata.org/dataset/cod-ab-mar)

```python
# set paths and url 
in_file =  'mar_admbnda_hcp_20230925_shp.zip'
in_url = f'https://data.humdata.org/dataset/7fb8f891-35f2-4bd5-8c75-5e7797ff1731/resource/26bb8834-1847-4ea2-bbab-e78c349f0aea/download/{in_file}'
out_path = f'{data_path}/data_raw/boundary/{in_file}'
os.makedirs(os.path.dirname(out_path), exist_ok=True)

# download the file if doesn't exist
if os.path.isfile(out_path):
    print(f'{in_file} already downloaded.')
else:
    print(f'Downloading {in_file}')
    download_file(in_url, out_path)

# unzip the file
if os.path.isfile(out_path):
    print(f'Unzipping {in_file}')
    unzip_file(out_path, os.path.dirname(out_path))
```
